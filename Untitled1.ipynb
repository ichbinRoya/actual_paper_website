```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from itertools import combinations

sns.set_theme(style="white")
df = sns.load_dataset('iris')

# All 6 pairs of the 4 measurements
pairs = list(combinations(['sepal_length', 'sepal_width',
                           'petal_length', 'petal_width'], 2))

# Set up the matplotlib figure
f, axes = plt.subplots(2, 3, figsize=(12, 8))

# Rotate the starting point around the cubehelix hue circle
for ax, (x_col, y_col), s in zip(axes.flat, pairs, np.linspace(0, 3, 6)):

    # Create a cubehelix colormap to use with kdeplot
    cmap = sns.cubehelix_palette(start=s, light=1, as_cmap=True)

    # Plot the density of this pair of measurements
    sns.kdeplot(
        data=df, x=x_col, y=y_col,
        cmap=cmap, fill=True,
        thresh=0, levels=15,
        ax=ax,
    )
    ax.set_title(f"{x_col} vs. {y_col}")

f.tight_layout()
plt.show()
```
